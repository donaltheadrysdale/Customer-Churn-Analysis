```python
#-------------------------------------------------
# STATISTICAL ANALYSIS: CHI SQUARE TEST OF INDEPENDENCE
# GENDER VS CHURN
# H0: THERE ARE NO SIGNIFICANT ASSOCIATIONS BETWEEN GENDER AND CHURN
# Ha: THERE IS SIGNIFICANT ASSOCIATION BETWEEN GENDER AND CHURN
# ALPHA: 0.05
#-------------------------------------------------
import scipy.stats as stats
from scipy.stats import chi2_contingency

# Create a table of gender versus customer churn
gender_table = pd.crosstab(df['Gender'], df['Exited'])

# Run the chi-square test
chi2, gender_p_value, dof, expected = chi2_contingency(gender_table)


print("Chi-square statistic:", chi2)
print("P-value:", gender_p_value)
print("Degrees of freedom:", dof)
print("Expected frequencies:\n", expected)
gender_table

#--------------------------------------------------
# P_VALUE < 0.05
# WE REJECT THE NULL HYPOTHESIS. THERE IS A SIGNIFICANT
# ASSOCIATION BETWEEN GENDER AND CHURN
#---------------------------------------------------
```

```python
#-------------------------------------------------
# STATISTICAL ANALYS: CRAMER V TEST
# QUESTION: HOW STRONG IS THE ASSOCICATION BETWEEN VARIABLES?
# SCALE 0 -> 1
#
# ES ≤ 0.2	The result is weak.
# 0.2 < ES ≤ 0.6	The result is moderate.
# ES > 0.6	The result is strong.
#-------------------------------------------------

import numpy as np

def v_cramers(contingency_table, chi2):
  rows, columns = contingency_table.shape
  n = contingency_table.to_numpy().sum()

  return np.sqrt(chi2 / (n * min(rows - 1, columns - 1))).round(4)

gender_es = v_cramers(gender_table, chi2)

# ------------------------------------------------
#
#
#-------------------------------------------------
```

```python
#-------------------------------------------------
# PRODUCT COUNT VS CHURN
# H0: THERE ARE NO SIGNIFICANT ASSOCIATIONS BETWEEN # OF PRODUCTS AND CHURN
# Ha: THERE IS SIGNIFICANT ASSOCIATION BETWEEN # OF PRODUCTS AND CHURN
# ALPHA: 0.05
#-------------------------------------------------

# Create a table of # of products versus customer churn
product_table = pd.crosstab(df['ProductGroup'], df['Exited'])

# Run the chi-square test
chi2, product_p_value, dof, expected = chi2_contingency(product_table)

print("Chi-square statistic:", chi2)
print("P-value:", product_p_value)
print("Degrees of freedom:", dof)
print("Expected frequencies:\n", expected)
product_table

#--------------------------------------------------
# P_VALUE < 0.05
# WE REJECT THE NULL HYPOTHESIS. THERE IS A SIGNIFICANT
# ASSOCIATION BETWEEN # OF PRODUCT AND CHURN
#---------------------------------------------------
```

```python
#-------------------------------------------------
# STATISTICAL ANALYS: CRAMER V TEST
# QUESTION: HOW STRONG IS THE ASSOCICATION BETWEEN VARIABLES?
# SCALE 0 -> 1
#
# ES ≤ 0.2	The result is weak.
# 0.2 < ES ≤ 0.6	The result is moderate.
# ES > 0.6	The result is strong.
#-------------------------------------------------


product_es = v_cramers(product_table, chi2)
product_es

# ------------------------------------------------
#
#
#-------------------------------------------------
```
