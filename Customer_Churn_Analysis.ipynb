```python
#-------------------------------------
# IMPORT DATASET
#-------------------------------------
import pandas as pd
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
pd.set_option('display.max_colwidth', None)

df = pd.read_csv("/content/Churn_Modelling.csv")
df = pd.DataFrame(df)
```

```python
#---------------------------------------
# INVESTIGATE DATASET
#---------------------------------------

print(df.info())
df.head()
print(df.describe())

```

```python
#---------------------------------------
# CHECK FOR DUPLICATES
#---------------------------------------
df.duplicated().sum()
```

```python
df['CreditGroup'] = pd.cut(
    df['CreditScore'],
    bins=[0, 580, 670, 740, 851],
    labels=['Poor', 'Fair', 'Good', 'Very Good'],
    right=False
)


df['AgeGroup'] = pd.cut(
    df['Age'],
    bins = [18,30,40,50,60,100],
    labels = ['Young Adult', 'Adult','Middle Age','Late Middle Age','Senior'],
    right = False
)

df['TenureGroup'] = pd.cut(
    df['Tenure'],
    bins = [0,2,5,8,10],
    labels = ['New Customer','Short Term Customer','Established Customer', 'Long Term Customer'],
    right = True
)

df['ProductGroup'] = pd.cut(
    df['NumOfProducts'],
    bins = [0,1,2,3,4],
    labels = ['Single Product Customer', 'Dual Product Customer', 'Triple Product Customer', 'Quadruple Product Customer'],
    right = True
)


df.head()
```

```python
#---------------------------------------
# QUESTION: WHAT IS THE CURRENT CHURN RATE?
#---------------------------------------
tot_customers = df.shape[0]

churned = df[df['Exited'] == 1]
total_churned = churned.shape[0]

churn_pct = round((total_churned / tot_customers)*100,2)
churn_pct #20.37

tot_nonchurned = tot_customers - total_churned



```

```python
#---------------------------------------
# CHURN PERCENTAGE BY GENDER
#---------------------------------------
gender_churn = df.groupby('Gender')['Exited'].agg(
    total_customers='count',
    churned_customers='sum',
    churn_rate='mean'
)

gender_churn['churn_rate'] = (
    gender_churn['churn_rate'] * 100
).round(2)

gender_churn
```

```python
#---------------------------------------
# CHURN PERCENTAGE BY CREDIT GROUP
#---------------------------------------
creditGroup = df.groupby('CreditGroup')['Exited'].agg(
    total_customers='count',
    churned_customers='sum',
    churn_rate='mean'
)

creditGroup['churn_rate'] = (
    creditGroup['churn_rate'] * 100
).round(2)

creditGroup
```
